## 1. Enquadramento Teórico Formal

### 1.1. Da Transformada de Fourier à Transformada de Laplace Bilateral

A **Transformada de Fourier em Tempo Contínuo (TF)** é uma ferramenta indispensável na análise espectral. No entanto, a sua aplicação requer que o sinal seja estritamente somável ou absolutamente integrável no sentido clássico:
$$\int_{-\infty}^{+\infty} |x(t)| \, dt < \infty$$
Sinais elementares de enorme relevância prática em Engenharia Electrotécnica — tais como o degrau unitário $u(t)$, rampas $t u(t)$, sinais sinusoidais com início temporal, ou respostas de circuitos e sistemas instáveis ($e^{at} u(t)$ com $a > 0$) - não cumprem esta condição de forma direta no domínio estrito das funções ordinárias.

Para contornar esta limitação matemática e unificar a análise de sistemas dinâmicos (estáveis e instáveis), introduz-se a **Transformada de Laplace Bilateral**, concebida através da multiplicação prévia do sinal temporal por uma exponencial real amortecedora $e^{-\sigma t}$:
$$X(s) = \mathcal{L}\{x(t)\} = \int_{-\infty}^{+\infty} \left[ x(t) e^{-\sigma t} \right] e^{-j\omega t} \, dt = \int_{-\infty}^{+\infty} x(t) e^{-(\sigma + j\omega) t} \, dt$$

Definindo a **variável complexa de frequência generalizada**:
$$s = \sigma + j\omega \in \mathbb{C}$$
obtém-se a definição canónica da **Transformada de Laplace Bilateral**:
$$\boxed{X(s) = \int_{-\infty}^{+\infty} x(t) e^{-st} \, dt}$$

* **Significado dos Componentes de $s$**:
  * $\sigma = \mathrm{Re}\{s\}$: taxa de variação exponencial (atenuação ou crescimento por unidade de tempo, em $\text{Np/s}$ ou $\text{rad/s}$).
  * $\omega = \mathrm{Im}\{s\}$: frequência angular de oscilação (em $\text{rad/s}$).
* **Relação com Fourier**: Quando a atenuação é nula ($\sigma = 0 \implies s = j\omega$), a Transformada de Laplace reduz-se à Transformada de Fourier convencional:
  $$\left. X(s) \right|_{s = j\omega} = \mathcal{F}\{x(t)\} = X(j\omega)$$

---

### 1.2. A Região de Convergência (ROC — *Region of Convergence*)

A integral de Laplace não converge para todos os valores de $s$ no plano complexo. O conjunto de pontos $s = \sigma + j\omega$ para os quais a integral converge absolutamente:
$$\int_{-\infty}^{+\infty} |x(t)| e^{-\sigma t} \, dt < \infty$$
denomina-se **Região de Convergência (ROC)**. 

> **Aviso**: A expressão analítica $X(s)$ **NÃO** determina univocamente o sinal temporal $x(t)$. Sinais completamente distintos no tempo podem partilhar a mesma expressão algébrica em $s$, diferindo apenas na sua ROC. Consequentemente, **especificar a ROC é obrigatório e indispensável**.

#### As 7 Propriedades Fundamentais da ROC:
1. **Geometria de Faixas Paralelas**: A condição de convergência depende exclusivamente de $\sigma = \mathrm{Re}\{s\}$. Logo, a ROC é constituída por faixas verticais ou semiplanos no plano complexo $s$, paralelos ao eixo imaginário $j\omega$.
2. **Ausência de Singularidades**: A ROC **nunca contém pólos** de $X(s)$, pois nos pólos a amplitude $|X(s)| \to \infty$.
3. **Sinais de Duração Finita**: Se $x(t)$ for um sinal de duração finita (isto é, $x(t) = 0$ para $t < t_1$ e $t > t_2$) e integrável, a ROC é **todo o plano-$s$**, exceto eventualmente $s = \infty$ (se contiver impulsos ou derivadas na origem) ou $s = -\infty$.
4. **Sinais à Direita (*Right-Sided* / Causais)**: Se $x(t)$ for um sinal que se inicia num instante finito $t_0$ ($x(t) = 0, \forall t < t_0$), a ROC é um **semiplano aberto à direita** do pólo com maior parte real:
   $$\text{ROC}: \mathrm{Re}\{s\} > \sigma_{\max}$$
5. **Sinais à Esquerda (*Left-Sided* / Anti-causais)**: Se $x(t)$ for um sinal que termina num instante finito $t_1$ ($x(t) = 0, \forall t > t_1$), a ROC é um **semiplano aberto à esquerda** do pólo com menor parte real:
   $$\text{ROC}: \mathrm{Re}\{s\} < \sigma_{\min}$$
6. **Sinais Bilaterais (*Two-Sided*)**: Se $x(t)$ tiver duração infinita em ambos os sentidos temporais ($t \to -\infty$ e $t \to +\infty$), a ROC é uma **faixa vertical aberta** compreendida entre dois pólos:
   $$\text{ROC}: \sigma_1 < \mathrm{Re}\{s\} < \sigma_2$$
   Se $\sigma_1 \ge \sigma_2$, a interseção é vazia e a Transformada de Laplace do sinal não existe.
7. **Existência da Transformada de Fourier**: O sinal $x(t)$ admite Transformada de Fourier convergente se e só se o **eixo imaginário $j\omega$ ($\mathrm{Re}\{s\} = 0$) estiver contido no interior da ROC**.

### 1.3. Diagrama Zero-Polar e Resposta no Tempo/Frequência

Para sistemas e sinais concentrados, a transformada $X(s)$ apresenta-se como uma função racional (quociente de dois polinómios em $s$):
$$X(s) = \frac{N(s)}{D(s)} = K \frac{(s - z_1)(s - z_2) \cdots (s - z_M)}{(s - p_1)(s - p_2) \cdots (s - p_N)}$$

* **Zeros ($z_i$)**: Valores de $s$ onde o numerador se anula ($N(z_i) = 0 \implies X(z_i) = 0$). Representam-se no plano-$s$ por pequenos círculos (**$o$**). Correspondem a frequências complexas bloqueadas pelo sistema.
* **Pólos ($p_k$)**: Raízes do polinómio característico do denominador ($D(p_k) = 0 \implies |X(p_k)| \to \infty$). Representam-se no plano-$s$ por cruzes (**$\times$**). Ditam os **modos naturais de oscilação e amortecimento** do sinal ou sistema ($e^{p_k t}$).
* **Efeito da Localização dos Pólos na Resposta Temporal**:
  * $\mathrm{Re}\{p_k\} < 0$ (Semiplano Esquerdo — SPE / LHP): Decaimento exponencial ($e^{-\alpha t}$), modos transitórios estáveis.
  * $\mathrm{Re}\{p_k\} > 0$ (Semiplano Direito — SPD / RHP): Crescimento exponencial ($e^{+\alpha t}$), modos instáveis.
  * $\mathrm{Re}\{p_k\} = 0$ (Eixo Imaginário $j\omega$): Oscilações puras não amortecidas ($\sin(\omega_0 t), \cos(\omega_0 t)$) ou componente contínua constante ($u(t)$ com pólo em $s=0$).
  * $\mathrm{Im}\{p_k\} \neq 0$: Presença de oscilações com frequência angular $\omega_d = |\mathrm{Im}\{p_k\}|$.

### 1.4. Metodologia de Cálculo da Transformada Inversa $\mathcal{L}^{-1}\{X(s)\}$

A inversão formal baseia-se na integral de contorno no plano complexo (Integral de Bromwich):
$$x(t) = \frac{1}{2\pi j} \int_{\sigma - j\infty}^{\sigma + j\infty} X(s) e^{st} \, ds$$
Na prática de Engenharia Electrotécnica, a inversão é efetuada através de **Expansão em Frações Simples** combinada com o conhecimento mandatório da ROC:

1. **Verificação do Grau dos Polinómios**:
   * Seja $M = \mathrm{grau}(N(s))$ e $N = \mathrm{grau}(D(s))$.
   * Se $M \ge N$ (Função Imprópria): Efetua-se previamente a **divisão polinomial** clássica:
     $$X(s) = Q(s) + \frac{R(s)}{D(s)}$$
     onde $Q(s)$ origina impulsos de Dirac e suas derivadas no tempo:
     $$\mathcal{L}^{-1}\{c_0\} = c_0 \delta(t), \quad \mathcal{L}^{-1}\{c_1 s\} = c_1 \delta'(t), \dots$$
     e o resto $\frac{R(s)}{D(s)}$ é estritamente próprio ($\mathrm{grau}(R) < N$).
2. **Decomposição em Frações Simples**:
   * **Pólos Reais Simples**: $\frac{R(s)}{D(s)} = \sum_{k=1}^N \frac{A_k}{s - p_k}$, com o resíduo calculado por $A_k = \lim_{s \to p_k} (s - p_k) X(s)$.
   * **Pólos Múltiplos de Ordem $m$**: Contribuição $\sum_{r=1}^m \frac{C_r}{(s - p_k)^r}$.
3. **Associação com a ROC para Determinar a Causalidade**:
   * Se a ROC estiver **à direita** do pólo $p_k$ ($\mathrm{Re}\{s\} > \mathrm{Re}\{p_k\}$):
     $$\mathcal{L}^{-1}\left\{ \frac{1}{s - p_k} \right\} = e^{p_k t} u(t) \quad (\text{Modo Causal})$$
   * Se a ROC estiver **à esquerda** do pólo $p_k$ ($\mathrm{Re}\{s\} < \mathrm{Re}\{p_k\}$):
     $$\mathcal{L}^{-1}\left\{ \frac{1}{s - p_k} \right\} = -e^{p_k t} u(-t) \quad (\text{Modo Anti-causal})$$
   * Para pólos de ordem 2 no semiplano esquerdo da ROC:
     $$\mathcal{L}^{-1}\left\{ \frac{1}{(s - p_k)^2} \right\} = -t e^{p_k t} u(-t)$$

### 1.5. Análise de Sistemas LIT Contínuos no Domínio de Laplace

Para um sistema Linear e Invariante no Tempo (LIT) contínuo com resposta impulsional $h(t)$:
$$y(t) = x(t) * h(t) \stackrel{\mathcal{L}}{\longleftrightarrow} Y(s) = H(s) X(s)$$

A **Função de Transferência** do sistema é definida por:
$$\boxed{H(s) = \frac{Y(s)}{X(s)} = \mathcal{L}\{h(t)\}}$$

* **Sistemas Descritos por EDLCC**:
  $$\sum_{k=0}^N a_k \frac{d^k y(t)}{dt^k} = \sum_{m=0}^M b_m \frac{d^m x(t)}{dt^m}$$
  Aplicando a propriedade da derivação no tempo com condições iniciais nulas ($\mathcal{L}\{\frac{d^n f}{dt^n}\} = s^n F(s)$):
  $$\left( \sum_{k=0}^N a_k s^k \right) Y(s) = \left( \sum_{m=0}^M b_m s^m \right) X(s) \implies H(s) = \frac{\sum_{m=0}^M b_m s^m}{\sum_{k=0}^N a_k s^k}$$

#### Critérios Formais de Causalidade e Estabilidade BIBO:
* **Causalidade**:
  * Um sistema LIT é causal se e só se $h(t) = 0$ para $t < 0$.
  * No domínio-$s$, para uma função racional, isto equivale a:
    1. A ROC de $H(s)$ é um semiplano aberto à direita do pólo com maior parte real: $\mathrm{Re}\{s\} > \max_k \mathrm{Re}\{p_k\}$.
    2. O grau do numerador não excede o grau do denominador ($M \le N$).
* **Estabilidade BIBO (*Bounded-Input, Bounded-Output*)**:
  * Um sistema LIT é estável BIBO se e só se a sua resposta impulsional for absolutamente integrável: $\int_{-\infty}^{+\infty} |h(t)| dt < \infty$.
  * No domínio-$s$, isto equivale rigorosamente a: **A ROC de $H(s)$ contém o eixo imaginário $j\omega$ ($\mathrm{Re}\{s\} = 0$)**.
* **Condição Unificada para Sistemas Causais e Estáveis**:
  * Para que um sistema LIT contínuo seja simultaneamente **causal** e **estável BIBO**, é condição necessária e suficiente que:
    $$\boxed{\text{Todos os pólos de } H(s) \text{ estejam estritamente situados no Semiplano Esquerdo (SPE): } \mathrm{Re}\{p_k\} < 0, \forall k}$$

```python
# ====================================================================
# CONFIGURAÇÃO DO AMBIENTE PYTHON & VISUALIZAÇÃO GRÁFICA PROFISSIONAL
# ====================================================================
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as signal
import sympy as sp
from sympy.integrals.transforms import laplace_transform, inverse_laplace_transform

# Configurações estéticas e tipográficas para Engenharia Electrotécnica
plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['figure.dpi'] = 120
plt.rcParams['font.size'] = 11
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.5
plt.rcParams['grid.linestyle'] = '--'
plt.rcParams['axes.axisbelow'] = True

def plot_pole_zero(zeros=None, poles=None, roc_type='right', roc_bounds=None, 
                   title='Diagrama Zero-Polar e ROC no Plano-s', 
                   xlim=(-5, 4), ylim=(-4, 4), ax=None):
    # Desenha o diagrama zero-polar com sombreamento profissional da ROC no plano-s.
    # zeros: lista de zeros complexos ou reais
    # poles: lista de pólos complexos ou reais
    # roc_type: 'right' (semiplano direito), 'left' (semiplano esquerdo), 'strip' (faixa)
    # roc_bounds: tuplo/lista com os limites reais da ROC
    if ax is None:
        fig, ax = plt.subplots(figsize=(6.5, 5.5), dpi=120)
    else:
        fig = ax.figure

    # Sombreamento da Região de Convergência (ROC)
    if roc_type == 'right' and roc_bounds is not None:
        ax.axvspan(roc_bounds[0], xlim[1], color='#38bdf8', alpha=0.28, 
                   label=f'ROC: $\\mathrm{{Re}}\\{{s\\}} > {roc_bounds[0]}$')
        ax.axvline(roc_bounds[0], color='#0284c7', linestyle='--', linewidth=1.8, 
                   label=f'Fronteira $\\sigma = {roc_bounds[0]}$')
    elif roc_type == 'left' and roc_bounds is not None:
        ax.axvspan(xlim[0], roc_bounds[0], color='#fb923c', alpha=0.28, 
                   label=f'ROC: $\\mathrm{{Re}}\\{{s\\}} < {roc_bounds[0]}$')
        ax.axvline(roc_bounds[0], color='#ea580c', linestyle='--', linewidth=1.8, 
                   label=f'Fronteira $\\sigma = {roc_bounds[0]}$')
    elif roc_type == 'strip' and roc_bounds is not None:
        ax.axvspan(roc_bounds[0], roc_bounds[1], color='#c084fc', alpha=0.28, 
                   label=f'ROC: ${roc_bounds[0]} < \\mathrm{{Re}}\\{{s\\}} < {roc_bounds[1]}$')
        ax.axvline(roc_bounds[0], color='#9333ea', linestyle='--', linewidth=1.8)
        ax.axvline(roc_bounds[1], color='#9333ea', linestyle='--', linewidth=1.8)

    # Eixos cartesianos no plano complexo
    ax.axhline(0, color='black', linewidth=1.2)
    ax.axvline(0, color='#1e293b', linewidth=1.6, linestyle='-', label='Eixo $j\\omega$ (Fourier)')

    # Marcação de Pólos (x a vermelho)
    if poles is not None and len(poles) > 0:
        poles = np.array(poles, dtype=complex)
        ax.scatter(poles.real, poles.imag, marker='x', s=130, color='#dc2626', 
                   linewidths=2.8, zorder=6, label='Pólos ($\\times$)')
        for p in poles:
            re_str = f"{p.real:.2g}" if abs(p.real) > 1e-4 else "0"
            im_str = f"{p.imag:+.2g}j" if abs(p.imag) > 1e-4 else ""
            coord_str = f"${re_str}{im_str}$".replace("+", "+").replace("+-", "-")
            ax.annotate(coord_str, (p.real, p.imag), textcoords="offset points", 
                        xytext=(8, 6), fontsize=10, weight='bold', color='#991b1b')

    # Marcação de Zeros (o a azul)
    if zeros is not None and len(zeros) > 0:
        zeros = np.array(zeros, dtype=complex)
        ax.scatter(zeros.real, zeros.imag, marker='o', s=110, facecolors='none', 
                   edgecolors='#2563eb', linewidths=2.5, zorder=6, label='Zeros ($o$)')
        for z in zeros:
            re_str = f"{z.real:.2g}" if abs(z.real) > 1e-4 else "0"
            im_str = f"{z.imag:+.2g}j" if abs(z.imag) > 1e-4 else ""
            coord_str = f"${re_str}{im_str}$".replace("+", "+").replace("+-", "-")
            ax.annotate(coord_str, (z.real, z.imag), textcoords="offset points", 
                        xytext=(8, 6), fontsize=10, weight='bold', color='#1e40af')

    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
    ax.set_xlabel('Eixo Real — Atenuação $\\sigma = \\mathrm{Re}\\{s\\}$ [Np/s]', fontsize=11, fontweight='bold')
    ax.set_ylabel('Eixo Imaginário — Frequência $j\\omega = j\\mathrm{Im}\\{s\\}$ [rad/s]', fontsize=11, fontweight='bold')
    ax.set_title(title, fontsize=12, fontweight='bold', pad=12)
    ax.legend(loc='upper right', framealpha=0.92, fontsize=9.5)
    return fig, ax

print("Ambiente Python configurado com sucesso! Bibliotecas NumPy, SciPy, SymPy e Matplotlib prontas para análise.")
```

#### Alínea 1. c) $$x(t) = e^{-at} u(t)$$

##### 1. Aplicação da Definição Integral
Pela definição da Transformada de Laplace Bilateral:
$$X(s) = \int_{-\infty}^{+\infty} x(t) e^{-st} \, dt$$
Substituindo $x(t) = e^{-at} u(t)$ (onde $u(t) = 1$ para $t \ge 0$ e $0$ para $t < 0$):
$$X(s) = \int_{0}^{+\infty} e^{-at} e^{-st} \, dt = \int_{0}^{+\infty} e^{-(s+a)t} \, dt$$

##### 2. Análise da Condição de Convergência (ROC)
Substituindo $s = \sigma + j\omega$:
$$e^{-(s+a)t} = e^{-(\sigma + \mathrm{Re}\{a\} + j(\omega + \mathrm{Im}\{a\}))t} = e^{-(\sigma + \mathrm{Re}\{a\})t} \cdot e^{-j(\omega + \mathrm{Im}\{a\})t}$$
A norma do integrando é:
$$\left| e^{-(s+a)t} \right| = e^{-(\sigma + \mathrm{Re}\{a\})t}$$
Para que a integral convirja no limite superior ($t \to +\infty$), o expoente real deve ser estritamente negativo:
$$-(\sigma + \mathrm{Re}\{a\}) < 0 \iff \sigma > -\mathrm{Re}\{a\} \iff \mathrm{Re}\{s\} > -\mathrm{Re}\{a\}$$
Assumindo $a \in \mathbb{R}$ (caso convencional em circuitos R-C / R-L):
$$\boxed{\text{ROC}: \mathrm{Re}\{s\} > -a}$$

##### 3. Avaliação da Primitiva
Sob a condição $\mathrm{Re}\{s\} > -a$:
$$X(s) = \left[ \frac{e^{-(s+a)t}}{-(s+a)} \right]_{t=0}^{t \to +\infty} = \lim_{t \to +\infty} \left( \frac{e^{-(s+a)t}}{-(s+a)} \right) - \left( \frac{e^0}{-(s+a)} \right) = 0 - \left( -\frac{1}{s+a} \right)$$
$$\boxed{X(s) = \frac{1}{s + a}, \quad \text{com } \mathrm{Re}\{s\} > -a}$$

##### 4. Discussão Físico-Matemática para Engenharia
* **Pólos e Zeros**: Possui um pólo simples em $s = -a$ e nenhum zero finito.
* **Geometria da ROC**: É um **semiplano aberto à direita** da linha vertical $\sigma = -a$. Isto reflete diretamente a natureza causal (à direita) do sinal $u(t)$.
* **Existência da Transformada de Fourier**: Se $a > 0$ (decaimento exponencial, sinal passivo amortecido), a reta vertical $\sigma = -a$ situa-se no semiplano esquerdo, pelo que o eixo $j\omega$ ($\sigma = 0$) pertence à ROC. Logo, a Transformada de Fourier converge: $X(j\omega) = \frac{1}{j\omega + a}$. Se $a \le 0$, o sinal cresce indefinidamente ou mantém-se constante, e o eixo $j\omega$ não está estritamente contido no interior da ROC.

#### Alínea 1. d) $$x(t) = u(3 - t)$$

##### 1. Natureza do Sinal
O sinal de degrau invertido $u(3-t)$ define-se como:
$$u(3-t) = \begin{cases} 1, & 3 - t \ge 0 \iff t \le 3 \\ 0, & t > 3 \end{cases}$$
Trata-se de um sinal **estritamente à esquerda** (*left-sided* / anti-causal deslocado), ativo desde $t = -\infty$ até ao instante de desativação $t = 3\text{ s}$.

##### 2. Resolução por Integração Direta
$$X(s) = \int_{-\infty}^{3} 1 \cdot e^{-st} \, dt = \left[ \frac{e^{-st}}{-s} \right]_{t=-\infty}^{t=3}$$
Analisando o comportamento no limite inferior ($t \to -\infty$):
$$e^{-st} = e^{-\sigma t} e^{-j\omega t}$$
Quando $t \to -\infty$, temos $-t \to +\infty$. Para que o termo $e^{-\sigma t}$ convirja para $0$ quando $t \to -\infty$, é imperativo que $-\sigma > 0 \iff \sigma < 0$, ou seja:
$$\mathrm{Re}\{s\} < 0$$
Aplicando os limites de integração sob esta condição:
$$X(s) = \frac{e^{-3s}}{-s} - \lim_{t \to -\infty} \frac{e^{-st}}{-s} = -\frac{e^{-3s}}{s} - 0$$
$$\boxed{X(s) = -\frac{e^{-3s}}{s}, \quad \text{com } \mathrm{Re}\{s\} < 0}$$

##### 3. Verificação Alternativa por Propriedades da Transformada
Podemos confirmar este resultado de forma elegante através das propriedades fundamentais:
1. **Sinal anti-causal de referência**: $x_0(t) = u(-t)$.
   $$\mathcal{L}\{u(-t)\} = \int_{-\infty}^0 e^{-st} dt = \left[ \frac{e^{-st}}{-s} \right]_{-\infty}^0 = -\frac{1}{s}, \quad \text{com } \mathrm{Re}\{s\} < 0$$
2. **Propriedade do Deslocamento Temporal**: $\mathcal{L}\{x_0(t - t_0)\} = e^{-s t_0} X_0(s)$.
3. Escrevendo $x(t) = u(3 - t) = u(-(t - 3)) = x_0(t - 3)$, com $t_0 = 3$:
   $$X(s) = e^{-3s} X_0(s) = e^{-3s} \left( -\frac{1}{s} \right) = -\frac{e^{-3s}}{s}, \quad \text{com } \mathrm{Re}\{s\} < 0$$
Ambos os métodos conduzem rigorosamente ao mesmo resultado!

##### 4. Discussão Físico-Matemática
* **Pólos e Zeros**: O termo tem uma singularidade essencial no infinito e um pólo em $s = 0$.
* **Geometria da ROC**: É um **semiplano aberto à esquerda** ($\mathrm{Re}\{s\} < 0$), o que é caraterístico de sinais que se estendem até $t = -\infty$.
* **Convergência de Fourier**: Como a fronteira da ROC é a reta $\sigma = 0$, o eixo imaginário $j\omega$ não está contido no interior da ROC. O sinal não admite Transformada de Fourier ordinária.

#### Alínea 1. e) $$x(t) = \sin(t) u(t)$$

##### 1. Decomposição de Euler
A função sinusoidal causal pode ser expressa em termos de exponenciais complexas através da identidade de Euler:
$$\sin(t) = \frac{e^{jt} - e^{-jt}}{2j}$$
Multiplicando por $u(t)$:
$$x(t) = \frac{1}{2j} e^{jt} u(t) - \frac{1}{2j} e^{-jt} u(t)$$

##### 2. Cálculo da Transformada por Linearidade
Recorrendo ao par deduzido na alínea 1.c), $\mathcal{L}\{e^{\alpha t} u(t)\} = \frac{1}{s - \alpha}$, com $\mathrm{Re}\{s\} > \mathrm{Re}\{\alpha\}$:
* Para a primeira componente ($\alpha_1 = j$):
  $$X_1(s) = \frac{1}{s - j}, \quad \text{com } \mathrm{Re}\{s\} > \mathrm{Re}\{j\} = 0$$
* Para a segunda componente ($\alpha_2 = -j$):
  $$X_2(s) = \frac{1}{s + j}, \quad \text{com } \mathrm{Re}\{s\} > \mathrm{Re}\{-j\} = 0$$

Aplicando a linearidade do operador de Laplace:
$$X(s) = \frac{1}{2j} X_1(s) - \frac{1}{2j} X_2(s) = \frac{1}{2j} \left( \frac{1}{s - j} - \frac{1}{s + j} \right)$$
Efetivando o denominador comum:
$$X(s) = \frac{1}{2j} \left( \frac{(s + j) - (s - j)}{(s - j)(s + j)} \right) = \frac{1}{2j} \left( \frac{2j}{s^2 - j^2} \right)$$
Como $j^2 = -1$:
$$\boxed{X(s) = \frac{1}{s^2 + 1}, \quad \text{com } \mathrm{Re}\{s\} > 0}$$

##### 3. Discussão Físico-Matemática
* **Pólos e Zeros**: Pólos complexos conjugados em $s = \pm j$ (raízes de $s^2 + 1 = 0$) e sem zeros finitos.
* **Geometria da ROC**: Semiplano aberto à direita $\mathrm{Re}\{s\} > 0$. Repare-se que os pólos situam-se exatamente **na fronteira** da ROC (no eixo imaginário $j\omega$).
* **Relação com Fourier**: Como os pólos se situam sobre o eixo $j\omega$ e a ROC não os engloba no seu interior, a Transformada de Fourier tradicional só pode ser formulada no sentido das distribuições (com recurso a deltas de Dirac ponderados nas frequências $\omega = \pm 1\text{ rad/s}$).

## Exercício 3

**Enunciado:**
> Calcule a transformada inversa de Laplace, $x(t) = \mathcal{L}^{-1}\{X(s)\}$ para cada uma das seguintes transformadas de Laplace, $X(s)$, e respetivas ROC:  
> **c)** $X(s) = \frac{2s + 1}{s + 2}, \quad \mathrm{Re}\{s\} > -2$  
> **d)** $X(s) = \frac{s^2 + 6s + 7}{s^2 + 3s + 2}, \quad \mathrm{Re}\{s\} > -1$  
> **f)** $X(s) = \frac{5s + 4}{s^2 + 2s + 1}, \quad \mathrm{Re}\{s\} < -1$  

---

### Resolução Analítica Detalhada Passo a Passo

---

#### Alínea 3. c) $$X(s) = \frac{2s + 1}{s + 2}, \quad \mathrm{Re}\{s\} > -2$$

##### 1. Análise dos Graus e Divisão Polinomial
Verificamos os graus dos polinómios:
* $\mathrm{grau}(N(s)) = 1$
* $\mathrm{grau}(D(s)) = 1$

Como $\mathrm{grau}(N) \ge \mathrm{grau}(D)$, a função racional **não é estritamente própria**. É obrigatório efetuar a divisão polinomial para isolar a parte inteira (que dará origem a termos de Dirac):
$$\frac{2s + 1}{s + 2} = \frac{2(s + 2) - 4 + 1}{s + 2} = \frac{2(s + 2) - 3}{s + 2} = 2 - \frac{3}{s + 2}$$

##### 2. Inversão Termo a Termo e Associação com a ROC
Temos agora duas parcelas:
1. **Termo constante $2$**:
   Pela propriedade fundamental da transformada de Laplace do impulso unitário de Dirac $\mathcal{L}\{\delta(t)\} = 1, \forall s$:
   $$\mathcal{L}^{-1}\{2\} = 2 \delta(t)$$
2. **Termo fracionário $-\frac{3}{s + 2}$**:
   Possui um pólo simples em $s = -2$.
   A ROC dada é $\mathrm{Re}\{s\} > -2$, ou seja, situa-se **à direita** do pólo $s = -2$. Logo, o modo temporal associado é estritamente **causal**:
   $$\mathcal{L}^{-1}\left\{ -\frac{3}{s + 2} \right\} = -3 e^{-2t} u(t)$$

##### 3. Expressão Analítica Final
Reunindo as parcelas no domínio do tempo:
$$\boxed{x(t) = 2 \delta(t) - 3 e^{-2t} u(t)}$$

> **Interpretação Física**: Este tipo de sinal surge tipicamente na resposta impulsional de circuitos elétricos com caminhos capacitivos diretos entre a entrada e a saída (como redes diferenciais ou filtros passa-alto reais), onde a aplicação de um impulso na entrada gera uma transferência instantânea de carga ($2\delta(t)$) seguida de uma descarga exponencial através das resistências do circuito ($-3e^{-2t}u(t)$).

---

#### Alínea 3. d) $$X(s) = \frac{s^2 + 6s + 7}{s^2 + 3s + 2}, \quad \mathrm{Re}\{s\} > -1$$

##### 1. Análise dos Graus e Divisão Polinomial
* $\mathrm{grau}(N(s)) = 2$
* $\mathrm{grau}(D(s)) = 2$

Como os graus são iguais, efetuamos a divisão de polinómios:
$$s^2 + 6s + 7 = 1 \cdot (s^2 + 3s + 2) + (3s + 5)$$
Logo:
$$X(s) = 1 + \frac{3s + 5}{s^2 + 3s + 2}$$

##### 2. Fatorização do Denominador e Expansão em Frações Simples
As raízes de $s^2 + 3s + 2 = 0$ são $s = -1$ e $s = -2$:
$$s^2 + 3s + 2 = (s + 1)(s + 2)$$
Decompomos a fração estritamente própria em frações parciais:
$$\frac{3s + 5}{(s + 1)(s + 2)} = \frac{A}{s + 1} + \frac{B}{s + 2}$$

Calculando os resíduos pelo **Método de Heaviside**:
$$A = \left. (s + 1) \frac{3s + 5}{(s + 1)(s + 2)} \right|_{s = -1} = \left. \frac{3s + 5}{s + 2} \right|_{s = -1} = \frac{3(-1) + 5}{-1 + 2} = \frac{2}{1} = 2$$
$$B = \left. (s + 2) \frac{3s + 5}{(s + 1)(s + 2)} \right|_{s = -2} = \left. \frac{3s + 5}{s + 1} \right|_{s = -2} = \frac{3(-2) + 5}{-2 + 1} = \frac{-1}{-1} = 1$$

Portanto:
$$X(s) = 1 + \frac{2}{s + 1} + \frac{1}{s + 2}$$

##### 3. Análise da ROC e Determinação do Sinal Temporal
A ROC especificada é:
$$\text{ROC}: \mathrm{Re}\{s\} > -1$$
Os pólos da função situam-se em $p_1 = -1$ e $p_2 = -2$.
Como $-1 > -2$, a condição $\mathrm{Re}\{s\} > -1$ coloca a ROC **à direita de AMBOS os pólos**.
Consequentemente, todas as parcelas fracionárias correspondem a termos temporais **causais** (ponderados por $u(t)$):
* $\mathcal{L}^{-1}\{1\} = \delta(t)$
* $\mathcal{L}^{-1}\left\{ \frac{2}{s + 1} \right\} = 2 e^{-t} u(t)$
* $\mathcal{L}^{-1}\left\{ \frac{1}{s + 2} \right\} = e^{-2t} u(t)$

Somando todas as parcelas:
$$\boxed{x(t) = \delta(t) + \left( 2 e^{-t} + e^{-2t} \right) u(t)}$$

---

#### Alínea 3. f) $$X(s) = \frac{5s + 4}{s^2 + 2s + 1}, \quad \mathrm{Re}\{s\} < -1$$

##### 1. Análise dos Graus e Fatorização
* $\mathrm{grau}(N(s)) = 1 < \mathrm{grau}(D(s)) = 2 \implies$ A função é estritamente própria (não contém deltas de Dirac).
* O denominador é um trinómio quadrado perfeito:
  $$s^2 + 2s + 1 = (s + 1)^2$$
Temos um **pólo duplo (ordem 2)** em $s = -1$.

##### 2. Expansão em Frações Parciais para Pólo Múltiplo
Procuramos uma expansão na forma:
$$\frac{5s + 4}{(s + 1)^2} = \frac{C_1}{s + 1} + \frac{C_2}{(s + 1)^2}$$
Podemos determinar os coeficientes por identificação algébrica direta:
$$5s + 4 = 5(s + 1) - 1 \implies \frac{5s + 4}{(s + 1)^2} = \frac{5(s + 1) - 1}{(s + 1)^2} = \frac{5}{s + 1} - \frac{1}{(s + 1)^2}$$
Logo:
$$C_1 = 5, \quad C_2 = -1$$

##### 3. Avaliação Crucial da ROC (Sinal Anti-Causal)
O enunciado estabelece:
$$\text{ROC}: \mathrm{Re}\{s\} < -1$$
> **Alerta Pedagógico de Extrema Importância**: A ROC situa-se **à esquerda** do pólo duplo $s = -1$. Consequentemente, o sinal temporal resultante é **estritamente anti-causal** (nulo para $t > 0$, ativo apenas para $t < 0$, acompanhado por $-u(-t)$).

Dedução formal dos pares anti-causais:
1. **Para o termo simples**:
   $$\mathcal{L}\{-e^{-at} u(-t)\} = \frac{1}{s + a}, \quad \text{com } \mathrm{Re}\{s\} < -a$$
   Para $a = 1$:
   $$\mathcal{L}^{-1}\left\{ \frac{5}{s + 1} \right\} = 5 \cdot \left[ -e^{-t} u(-t) \right] = -5 e^{-t} u(-t)$$
2. **Para o termo de ordem 2**:
   Aplicando a propriedade da diferenciação no domínio da frequência:
   $$\mathcal{L}\{t f(t)\} = -\frac{d}{ds} F(s)$$
   Tomando $f(t) = -e^{-t} u(-t) \implies F(s) = \frac{1}{s + 1}$ com $\mathrm{Re}\{s\} < -1$:
   $$\mathcal{L}\{t (-e^{-t} u(-t))\} = -\frac{d}{ds} \left( \frac{1}{s + 1} \right) = -\left( -\frac{1}{(s + 1)^2} \right) = \frac{1}{(s + 1)^2}$$
   Multiplicando ambos os membros por $-1$:
   $$\mathcal{L}\{-t e^{-t} u(-t)\} = -\frac{1}{(s + 1)^2} \implies \mathcal{L}\{t e^{-t} u(-t)\} = -\frac{1}{(s + 1)^2}$$
   Logo:
   $$\mathcal{L}^{-1}\left\{ -\frac{1}{(s + 1)^2} \right\} = t e^{-t} u(-t)$$

##### 4. Expressão Temporal Final
Somando as duas componentes:
$$x(t) = -5 e^{-t} u(-t) + t e^{-t} u(-t) = (t - 5) e^{-t} u(-t)$$
Podemos também escrever na forma equivalente:
$$\boxed{x(t) = -(5 - t) e^{-t} u(-t)}$$

*Verificação da consistência*: Para $t < 0$, à medida que $t \to -\infty$, o termo exponencial $e^{-t} = e^{|t|}$ cresce exponencialmente, o que é plenamente compatível com a ROC $\mathrm{Re}\{s\} < -1$, uma vez que para valores de $\sigma < -1$, o fator $e^{-\sigma t}$ suprime esse crescimento temporal, assegurando a integrabilidade da transformada.

## Exercício 6
### Análise Completa de Sistema LIT: Função de Transferência, Causalidade, Estabilidade e Resposta a Sinais Arbitrários

**Enunciado:**
> A saída de um sistema LIT é $$y(t) = 2e^{-3t}u(t)$$ quando a entrada é $$x(t) = u(t)$$.  
> **a)** Obtenha a função de transferência do sistema, $H(s)$, caraterizando-o relativamente à ROC, causalidade e estabilidade.  
> **b)** Determine a resposta impulsional do sistema, $h(t)$.  
> **c)** Determine a saída, $y(t)$, quando a entrada é $x(t) = e^{-t}u(t)$.  

---

### Resolução Analítica Detalhada Passo a Passo

---

#### Alínea 6. a) Obtenção de $$H(s)$$, ROC, Causalidade e Estabilidade

##### 1. Transformada da Entrada e da Saída
1. **Sinal de Entrada**:
   $$x(t) = u(t) \iff X(s) = \frac{1}{s}, \quad \text{com } \text{ROC}_x: \mathrm{Re}\{s\} > 0$$
2. **Sinal de Saída**:
   $$y(t) = 2e^{-3t} u(t) \iff Y(s) = \frac{2}{s + 3}, \quad \text{com } \text{ROC}_y: \mathrm{Re}\{s\} > -3$$

##### 2. Dedução da Função de Transferência $$H(s)$$
Para qualquer sistema LIT contínuo, a convolução temporal $y(t) = x(t) * h(t)$ transforma-se no produto algébrico $Y(s) = H(s) X(s)$.
Isolando $H(s)$:
$$H(s) = \frac{Y(s)}{X(s)} = \frac{\frac{2}{s + 3}}{\frac{1}{s}} = \frac{2s}{s + 3}$$
$$\boxed{H(s) = \frac{2s}{s + 3}}$$

##### 3. Zeros e Pólos de $$H(s)$$
* **Zero**: $N(s) = 2s = 0 \implies z_1 = 0$ (zero na origem do plano-$s$).
* **Pólo**: $D(s) = s + 3 = 0 \implies p_1 = -3$ (pólo real no semiplano esquerdo).

##### 4. Caracterização da Causalidade e da ROC
* **Causalidade**: O sistema, ao ser excitado por uma entrada causal $x(t) = u(t)$ ($x(t) = 0, \forall t < 0$), produz uma resposta estritamente causal $y(t) = 2e^{-3t}u(t)$ ($y(t) = 0, \forall t < 0$). Estando o sistema inicialmente em repouso (relaxado), o sistema é **CAUSAL**.
* **Determinação da ROC de $H(s)$**: Para um sistema LIT contínuo causal descrito por uma função de transferência racional, a ROC é obrigatoriamente o semiplano aberto à direita do pólo com maior parte real. Como o único pólo é $p_1 = -3$:
  $$\boxed{\text{ROC}_h: \mathrm{Re}\{s\} > -3}$$

##### 5. Caracterização da Estabilidade BIBO
* **Critério de Estabilidade**: Um sistema LIT contínuo é estável BIBO (*Bounded-Input, Bounded-Output*) se e só se a sua resposta impulsional for absolutamente integrável ($\int_{-\infty}^{+\infty} |h(t)| dt < \infty$), o que equivale a que a **ROC de $H(s)$ contenha o eixo imaginário $j\omega$ ($\mathrm{Re}\{s\} = 0$)**.
* Como a ROC é $\mathrm{Re}\{s\} > -3$ e $-3 < 0$, o eixo imaginário $\sigma = 0$ está garantidamente contido no interior da ROC.
* Equivalentemente, tratando-se de um sistema causal, todos os seus pólos situam-se estritamente no semiplano esquerdo ($\mathrm{Re}\{p_1\} = -3 < 0$).
* **Conclusão**: O sistema é **ESTÁVEL (BIBO)**.

---

#### Alínea 6. b) Determinação da Resposta Impulsional $$h(t)$$

Podemos resolver esta questão através de **dois métodos analíticos independentes**, o que permite aos estudantes validar a consistência física do modelo:

##### Método 1: Inversão Direta da Função de Transferência $$H(s)$$
Temos:
$$H(s) = \frac{2s}{s + 3}, \quad \text{com } \mathrm{Re}\{s\} > -3$$
Como $\mathrm{grau}(N) = \mathrm{grau}(D) = 1$, realizamos a divisão polinomial:
$$H(s) = \frac{2(s + 3) - 6}{s + 3} = 2 - \frac{6}{s + 3}$$
Invertendo termo a termo para o domínio temporal, atendendo à ROC causal $\mathrm{Re}\{s\} > -3$:
$$\mathcal{L}^{-1}\{2\} = 2 \delta(t)$$
$$\mathcal{L}^{-1}\left\{ -\frac{6}{s + 3} \right\} = -6 e^{-3t} u(t)$$
Logo:
$$\boxed{h(t) = 2 \delta(t) - 6 e^{-3t} u(t)}$$

##### Método 2: Relação entre a Resposta ao Degrau e a Resposta ao Impulso
Por definição, quando a entrada é o degrau unitário $x(t) = u(t)$, a saída do sistema é a sua **resposta ao escalão**, habitualmente designada por $s(t)$:
$$s(t) = y(t) = 2e^{-3t} u(t)$$
Dado que o impulso unitário é a derivada generalizada do degrau unitário, $\delta(t) = \frac{d}{dt} u(t)$, e sendo o sistema linear e invariante no tempo, a resposta ao impulso $h(t)$ é a **derivada temporal da resposta ao escalão**:
$$h(t) = \frac{d}{dt} s(t) = \frac{d}{dt} \left[ 2e^{-3t} u(t) \right]$$
Aplicando a regra de derivação do produto de funções (no sentido das distribuições):
$$h(t) = 2 \left[ \frac{d}{dt}\left(e^{-3t}\right) \cdot u(t) + e^{-3t} \cdot \frac{d}{dt} u(t) \right]$$
$$h(t) = 2 \left[ -3 e^{-3t} u(t) + e^{-3t} \delta(t) \right]$$
Pela propriedade da filtragem/amostragem do impulso de Dirac, $f(t) \delta(t) = f(0) \delta(t)$:
$$e^{-3t} \delta(t) = e^{-3(0)} \delta(t) = 1 \cdot \delta(t) = \delta(t)$$
Substituindo:
$$h(t) = -6 e^{-3t} u(t) + 2 \delta(t) = 2 \delta(t) - 6 e^{-3t} u(t)$$
Obtém-se rigorosamente a mesma expressão!

---

#### Alínea 6. c) Saída $$y(t)$$ para Entrada $$x(t) = e^{-t} u(t)$$

##### 1. Transformada da Nova Entrada
$$x(t) = e^{-t} u(t) \iff X(s) = \frac{1}{s + 1}, \quad \text{com } \text{ROC}_x: \mathrm{Re}\{s\} > -1$$

##### 2. Cálculo da Nova Saída no Domínio de Laplace
A transformada da saída é dada por:
$$Y(s) = H(s) X(s) = \left( \frac{2s}{s + 3} \right) \left( \frac{1}{s + 1} \right) = \frac{2s}{(s + 1)(s + 3)}$$

A ROC de $Y(s)$ contém pelo menos a interseção das ROCs de $H(s)$ e $X(s)$:
$$\text{ROC}_y = \text{ROC}_h \cap \text{ROC}_x = \{ \mathrm{Re}\{s\} > -3 \} \cap \{ \mathrm{Re}\{s\} > -1 \} = \{ \mathrm{Re}\{s\} > -1 \}$$

##### 3. Decomposição em Frações Simples
Como $\mathrm{grau}(N) = 1 < \mathrm{grau}(D) = 2$, $Y(s)$ é estritamente própria:
$$\frac{2s}{(s + 1)(s + 3)} = \frac{A}{s + 1} + \frac{B}{s + 3}$$
Calculando os resíduos:
$$A = \left. \frac{2s}{s + 3} \right|_{s = -1} = \frac{2(-1)}{-1 + 3} = \frac{-2}{2} = -1$$
$$B = \left. \frac{2s}{s + 1} \right|_{s = -3} = \frac{2(-3)}{-3 + 1} = \frac{-6}{-2} = 3$$
Portanto:
$$Y(s) = -\frac{1}{s + 1} + \frac{3}{s + 3}, \quad \text{com } \mathrm{Re}\{s\} > -1$$

##### 4. Transformada Inversa e Resposta Temporal
Sendo a ROC $\mathrm{Re}\{s\} > -1$ (à direita de ambos os pólos $s = -1$ e $s = -3$), ambas as componentes são causais:
$$y(t) = \left( -e^{-t} + 3e^{-3t} \right) u(t)$$
Reordenando:
$$\boxed{y(t) = \left( 3 e^{-3t} - e^{-t} \right) u(t)}$$

##### 5. Verificação Rigorosa pelos Teoremas dos Limites
* **Teorema do Valor Inicial**:
  $$y(0^+) = \lim_{s \to \infty} s Y(s) = \lim_{s \to \infty} \frac{2s^2}{(s+1)(s+3)} = \lim_{s \to \infty} \frac{2s^2}{s^2 + 4s + 3} = 2$$
  No domínio temporal: $y(0^+) = 3 e^0 - e^0 = 3 - 1 = 2$. **Perfeito!**
* **Teorema do Valor Final**:
  Dado que todos os pólos de $s Y(s)$ estão no semiplano esquerdo estrito:
  $$\lim_{t \to \infty} y(t) = \lim_{s \to 0} s Y(s) = \lim_{s \to 0} \frac{2s^2}{(s+1)(s+3)} = 0$$
  No domínio temporal: $\lim_{t \to \infty} (3e^{-3t} - e^{-t}) = 0$. **Perfeito!**

```python
# ====================================================================
# SIMULAÇÃO COMPUTACIONAL — Exercício 6
# ====================================================================

# 1. Diagrama Zero-Polar e Respostas Temporais do Sistema
fig, axs = plt.subplots(2, 2, figsize=(15, 9), dpi=120)

# Diagrama Zero-Polar de H(s) = 2s / (s + 3)
plot_pole_zero(zeros=[0], poles=[-3], roc_type='right', roc_bounds=[-3],
               title='Função de Transferência $H(s) = \\frac{2s}{s + 3}$ (Causal e Estável)',
               xlim=(-5, 3), ylim=(-3, 3), ax=axs[0, 0])

# Resposta ao Degrau dada: y(t) = 2*exp(-3t)*u(t)
t_sim = np.linspace(-0.5, 3.5, 1000)
y_step = np.where(t_sim >= 0, 2*np.exp(-3*t_sim), 0.0)
axs[0, 1].plot(t_sim, y_step, color='#0284c7', linewidth=2.5, label='Resposta ao Degrau: $s(t) = 2e^{-3t}u(t)$')
axs[0, 1].set_title('a) Resposta ao Degrau Unitário $x(t) = u(t)$', fontweight='bold')
axs[0, 1].set_xlabel('Tempo $t$ [s]')
axs[0, 1].set_ylabel('Amplitude')
axs[0, 1].legend()

# Resposta Impulsional: h(t) = 2*delta(t) - 6*exp(-3t)*u(t)
h_cont = np.where(t_sim > 0, -6*np.exp(-3*t_sim), 0.0)
axs[1, 0].plot(t_sim, h_cont, color='#ea580c', linewidth=2.5, label='Modo Transitório: $-6e^{-3t}u(t)$')
axs[1, 0].annotate('', xy=(0, 2.0), xytext=(0, 0),
                   arrowprops=dict(arrowstyle="->", color='#dc2626', lw=3.0))
axs[1, 0].scatter([0], [2.0], color='#dc2626', s=90, zorder=5, label='Impulso de Dirac: $2\\delta(t)$')
axs[1, 0].set_title('b) Resposta Impulsional $h(t) = 2\\delta(t) - 6e^{-3t}u(t)$', fontweight='bold')
axs[1, 0].set_xlabel('Tempo $t$ [s]')
axs[1, 0].set_ylabel('Amplitude')
axs[1, 0].set_ylim(-6.5, 2.5)
axs[1, 0].legend()

# Resposta a x(t) = exp(-t)*u(t): y(t) = (3*exp(-3t) - exp(-t))*u(t)
# Comparação entre a expressão teórica e a simulação com scipy.signal
num_H = [2, 0]    # 2s
den_H = [1, 3]    # s + 3
sys_lti = signal.lti(num_H, den_H)

t_pos = np.linspace(0, 3.5, 500)
u_in = np.exp(-t_pos)
t_out, y_out_scipy, _ = signal.lsim(sys_lti, U=u_in, T=t_pos)
y_teorico = (3*np.exp(-3*t_pos) - np.exp(-t_pos))

axs[1, 1].plot(t_pos, y_teorico, color='#16a34a', linewidth=3.0, label='Analítico: $(3e^{-3t} - e^{-t})u(t)$')
axs[1, 1].plot(t_out, y_out_scipy, color='#1e293b', linestyle='--', linewidth=2.0, label='Simulação SciPy (`lsim`)')
axs[1, 1].set_title('c) Resposta à Excitação $x(t) = e^{-t} u(t)$', fontweight='bold')
axs[1, 1].set_xlabel('Tempo $t$ [s]')
axs[1, 1].set_ylabel('Amplitude')
axs[1, 1].legend()

plt.tight_layout()
plt.show()

print("Simulação do Exercício 6 concluída com total concordância teórica e computacional.")
```

## Exercício 13
### Sistemas LIT Regidos por Equações Diferenciais Lineares de Coeficientes Constantes (EDLCC)

**Enunciado:**
> Considere um sistema LIT contínuo, causal, descrito pela equação diferencial linear de coeficientes constantes:  
> $$\ddot{y}(t) + 3 \dot{y}(t) + 2y(t) = 2 \dot{x}(t) + 3x(t)$$  
> Esboce o diagrama zero-polar no plano-$s$ e determine a resposta impulsional do sistema, $h(t)$.  

---

### Resolução Analítica Detalhada Passo a Passo

---

#### 1. Transformada de Laplace da Equação Diferencial
Para caracterizar a resposta do sistema LIT (representação relaxada / condições iniciais nulas no instante $t = 0^-$):
$$y(0^-) = 0, \quad \dot{y}(0^-) = 0, \quad x(0^-) = 0$$

Aplicando a propriedade da diferenciação no tempo da Transformada de Laplace:
$$\mathcal{L}\left\{ \frac{d^n f(t)}{dt^n} \right\} = s^n F(s)$$
Transformamos termo a termo a equação diferencial:
$$\mathcal{L}\{\ddot{y}(t)\} = s^2 Y(s)$$
$$\mathcal{L}\{3\dot{y}(t)\} = 3s Y(s)$$
$$\mathcal{L}\{2y(t)\} = 2 Y(s)$$
$$\mathcal{L}\{2\dot{x}(t)\} = 2s X(s)$$
$$\mathcal{L}\{3x(t)\} = 3 X(s)$$

Substituindo na equação:
$$s^2 Y(s) + 3s Y(s) + 2 Y(s) = 2s X(s) + 3 X(s)$$
Agrupando os termos:
$$(s^2 + 3s + 2) Y(s) = (2s + 3) X(s)$$

---

#### 2. Determinação da Função de Transferência $$H(s)$$
A Função de Transferência é a razão entre a transformada da saída e a da entrada:
$$H(s) = \frac{Y(s)}{X(s)} = \frac{2s + 3}{s^2 + 3s + 2}$$

Fatorizando o polinómio característico do denominador $s^2 + 3s + 2$:
$$s^2 + 3s + 2 = (s + 1)(s + 2)$$
Obtém-se a forma canónica fatorizada:
$$\boxed{H(s) = \frac{2s + 3}{(s + 1)(s + 2)}}$$

---

#### 3. Zeros, Pólos, Diagrama Zero-Polar e ROC
* **Zeros ($z_i$)**:
  Raízes do numerador:
  $$2s + 3 = 0 \implies z_1 = -\frac{3}{2} = -1.5$$
  (Um zero real simples no semiplano esquerdo).
* **Pólos ($p_k$)**:
  Raízes do denominador:
  $$(s + 1)(s + 2) = 0 \implies p_1 = -1, \quad p_2 = -2$$
  (Dois pólos reais simples no semiplano esquerdo).
* **Região de Convergência (ROC)**:
  O enunciado estabelece expressamente que o sistema é **CAUSAL**.
  Para um sistema causal racional, a ROC é o semiplano aberto delimitado à direita pelo pólo com a maior parte real ($\max(-1, -2) = -1$):
  $$\boxed{\text{ROC}_h: \mathrm{Re}\{s\} > -1}$$
* **Estabilidade BIBO**:
  Ambos os pólos situam-se estritamente no semiplano esquerdo ($\mathrm{Re}\{p_1\} = -1 < 0$ e $\mathrm{Re}\{p_2\} = -2 < 0$).
  Como a ROC é $\mathrm{Re}\{s\} > -1$, ela inclui o eixo imaginário $j\omega$ ($\mathrm{Re}\{s\} = 0$).
  Consequentemente, o sistema é **ESTÁVEL BIBO**.

---

#### 4. Cálculo da Resposta Impulsional $$h(t)$$
Sendo $\mathrm{grau}(N) = 1 < \mathrm{grau}(D) = 2$, a função de transferência é estritamente própria (não possui termo constante e, portanto, $h(t)$ não contém impulsos de Dirac na origem).

Efetuamos a **decomposição em frações simples**:
$$H(s) = \frac{2s + 3}{(s + 1)(s + 2)} = \frac{A}{s + 1} + \frac{B}{s + 2}$$

Calculando os resíduos pelo Método de Heaviside:
$$A = \left. \frac{2s + 3}{s + 2} \right|_{s = -1} = \frac{2(-1) + 3}{-1 + 2} = \frac{1}{1} = 1$$
$$B = \left. \frac{2s + 3}{s + 1} \right|_{s = -2} = \frac{2(-2) + 3}{-2 + 1} = \frac{-1}{-1} = 1$$

Portanto:
$$H(s) = \frac{1}{s + 1} + \frac{1}{s + 2}, \quad \text{com } \mathrm{Re}\{s\} > -1$$

Invertendo para o domínio temporal, atendendo à causalidade garantida pela ROC ($\mathrm{Re}\{s\} > -1$):
$$\boxed{h(t) = \left( e^{-t} + e^{-2t} \right) u(t)}$$

#### 5. Interpretação em Engenharia Electrotécnica
* **Modos Naturais do Sistema**: O sistema exibe dois modos transitórios de decaimento puramente exponencial:
  1. Modo Rápido: $e^{-2t}$, com constante de tempo $\tau_2 = \frac{1}{2} = 0.5\text{ s}$.
  2. Modo Lento / Dominante: $e^{-t}$, com constante de tempo $\tau_1 = \frac{1}{1} = 1.0\text{ s}$.
  O pólo em $s = -1$, por estar mais próximo do eixo $j\omega$, é o **pólo dominante** do sistema, determinando a velocidade de assentamento em regime transitório.
* **Verificação do Valor Inicial**:
  $$h(0^+) = e^0 + e^0 = 1 + 1 = 2$$
  Pelo Teorema do Valor Inicial:
  $$\lim_{s \to \infty} s H(s) = \lim_{s \to \infty} \frac{s(2s + 3)}{s^2 + 3s + 2} = \lim_{s \to \infty} \frac{2s^2 + 3s}{s^2 + 3s + 2} = 2$$
  A confirmação é imediata e rigorosa!

```python
# ====================================================================
# SIMULAÇÃO COMPUTACIONAL — Exercício 13
# ====================================================================

# 1. Configuração do Sistema LTI no SciPy
num = [2, 3]       # 2s + 3
den = [1, 3, 2]    # s^2 + 3s + 2
sys_edl = signal.lti(num, den)

# Base temporal
t_imp = np.linspace(0, 5, 1000)

# Simulação da Resposta Impulsional com scipy.signal.impulse
t_sim, h_sim = signal.impulse(sys_edl, T=t_imp)

# Expressão Teórica Analítica
h_teorico = np.exp(-t_imp) + np.exp(-2*t_imp)
modo_lento = np.exp(-t_imp)
modo_rapido = np.exp(-2*t_imp)

# 2. Representação Gráfica: Diagrama Zero-Polar e Resposta Impulsional
fig, axs = plt.subplots(1, 2, figsize=(15, 6), dpi=120)

# Diagrama Zero-Polar
plot_pole_zero(zeros=[-1.5], poles=[-1, -2], roc_type='right', roc_bounds=[-1],
               title='Diagrama Zero-Polar de $H(s) = \\frac{2s + 3}{(s + 1)(s + 2)}$',
               xlim=(-4, 2), ylim=(-3, 3), ax=axs[0])

# Resposta Impulsional e Decomposição de Modos
axs[1].plot(t_sim, h_sim, color='#1e293b', linestyle='--', linewidth=3.0, label='Simulação Numérica (`impulse`)')
axs[1].plot(t_imp, h_teorico, color='#2563eb', linewidth=2.0, label='Analítico Total: $h(t) = (e^{-t} + e^{-2t})u(t)$')
axs[1].plot(t_imp, modo_lento, color='#16a34a', linestyle=':', linewidth=2.0, label='Modo Lento Dominante: $e^{-t}$')
axs[1].plot(t_imp, modo_rapido, color='#dc2626', linestyle=':', linewidth=2.0, label='Modo Rápido: $e^{-2t}$')

axs[1].scatter([0], [2.0], color='#2563eb', s=80, zorder=5, label='Valor Inicial: $h(0^+) = 2.0$')
axs[1].set_title('Resposta Impulsional $h(t)$ e Modos Naturais', fontweight='bold')
axs[1].set_xlabel('Tempo $t$ [s]')
axs[1].set_ylabel('Amplitude $h(t)$')
axs[1].legend()

plt.tight_layout()
plt.show()

print("Exercício 13 simulado com sucesso: validação perfeita dos pólos, zero e resposta impulsional.")
```

## 4. Recomendações e Conclusões Pedagógicas do Docente

Estimados estudantes de Engenharia Electrotécnica, ao concluirmos a análise detalhada desta folha prática, destacam-se quatro princípios basilares que devem reter para as avaliações e para a vossa futura atividade profissional:

1. **A Indissociabilidade entre a Expressão Algébrica e a ROC**:
   * Nunca apresentem uma Transformada de Laplace ou Função de Transferência sem declarar expressamente a respetiva Região de Convergência.
   * Conforme demonstrado nos Exercícios 1.d), 2.b), 2.f) e 3.f), a mesma expressão algébrica pode representar sinais físicos completamente distintos (causais, anti-causais ou bilaterais). É a ROC que dita a causalidade e a estabilidade!

2. **O Risco da Divisão de Polinómios Omitida**:
   * Sempre que o grau do numerador for superior ou igual ao grau do denominador ($\mathrm{grau}(N) \ge \mathrm{grau}(D)$), a função racional **não é estritamente própria**.
   * A omissão da divisão polinomial prévia leva à perda irreversível dos termos de impulso de Dirac $\delta(t)$ e das suas derivadas $\delta'(t)$, como ficou patente nos Exercícios 3.c), 3.d) e 6.b). Em circuitos elétricos, estes termos traduzem transferências instantâneas de carga capacitiva ou saltos abruptos de corrente indutiva.

3. **Causalidade vs. Estabilidade BIBO**:
   * A estabilidade BIBO em tempo contínuo exige rigorosamente que a **ROC englobe o eixo imaginário $j\omega$ ($\mathrm{Re}\{s\} = 0$)**.
   * A causalidade exige que a ROC seja um semiplano direito.
   * Da conjugação destas duas exigências resulta o critério fundamental para sistemas práticos: **todos os pólos do sistema devem localizar-se estritamente no Semiplano Esquerdo (SPE / LHP)**.

4. **Transição para Teoria de Controlo e Circuitos no Domínio da Frequência**:
   * O diagrama zero-polar constitui o mapa genético de qualquer circuito ou sistema dinâmico.
   * A resposta em frequência em regime sinusoidal permanente obtém-se diretamente avaliando $H(s)$ ao longo do eixo imaginário: $H(j\omega) = \left. H(s) \right|_{s = j\omega}$.
   * Os pólos mais próximos do eixo $j\omega$ (pólos dominantes) ditarão a velocidade de resposta transitória e os sobre-sinais em malha fechada.
